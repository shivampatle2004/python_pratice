### Default Arguments

Default arguments allow you to assign a default value to an argument in a function definition. If the caller doesn't provide a value for that argument, the default value is used. This makes functions more flexible and can reduce the number of arguments needed in common use cases.

```python
def squ(a=1,b=1):
  return a**b
```

The `squ` function calculates `a` raised to the power of `b`. Both `a` and `b` have default values of `1`. This means if you call `squ()` without any arguments, it will return `1**1 = 1`. If you provide only `a`, `b` will still be `1` by default (e.g., `squ(5)` returns `5**1 = 5`).

### Positional Arguments

When you call a function, arguments are passed to parameters in the order they are defined. These are called positional arguments. The first argument in the call maps to the first parameter in the definition, and so on.

```python
squ(2,3)#2=a and 3=b
```

When calling `squ(2,3)`, `2` is assigned to `a` (the first parameter) and `3` is assigned to `b` (the second parameter). The function then calculates `2**3`, which is 8.

### Keyword Arguments

Keyword arguments allow you to specify the argument by its parameter name when calling a function. This makes the code more readable and allows you to pass arguments in any order, as long as you name them.

```python
squ(b=3,a=2)
```

Here, we explicitly use keyword arguments `b=3` and `a=2`. This allows us to pass arguments in any order, and the function correctly assigns `2` to `a` and `3` to `b`, resulting in `2**3 = 8`. Keyword arguments make your code more readable by clearly indicating what each argument represents.

### Arbitrary Arguments (`*args` and `**kwargs`)

Sometimes you don't know in advance how many arguments a function will need to accept. Python provides special syntaxes for handling an arbitrary number of arguments:

*   **`*args` (Arbitrary Positional Arguments):** Allows a function to accept any number of positional arguments. These arguments are packed into a tuple.
*   **`**kwargs` (Arbitrary Keyword Arguments):** Allows a function to accept any number of keyword arguments. These arguments are packed into a dictionary.

```python
def f1(*salll):
  for i in salll:
    print(i)
f1(1,2,3,4,5,6,7,8)
```

The `f1(*salll)` function uses `*args`. This allows `f1` to accept an arbitrary number of positional arguments. All arguments passed are collected into a tuple named `salll`. The function then iterates through this tuple and prints each element.

```python
def f2(**sal):
  for (k,v) in sal.items():
    print(k,"->",v)
f2(india="delhi",pakistan="kabul")
```

The `f2(**sal)` function uses `**kwargs`. This allows `f2` to accept an arbitrary number of keyword arguments. These arguments are collected into a dictionary named `sal`, where the keywords become dictionary keys and their values become dictionary values. The function then iterates through this dictionary and prints each key-value pair.

Write a Python function to check whether a number is perfect or not.
A Perfect number is a number that is half the sum of all of its positive divisors (including itself).

Example :

The first perfect number is 6, because 1, 2, and 3 are its proper positive divisors, and 1 + 2 + 3 = 6.
Equivalently, the number 6 is equal to half the sum of all its positive divisors: ( 1 + 2 + 3 + 6 ) / 2 = 6.

The next perfect number is 28 = 1 + 2 + 4 + 7 + 14. This is followed by the perfect numbers 496 and 8128.

```python
def f1(n):
 s=0
 for i in range(1,n):
  if n%i==0 :
    s=s+i
    # print(s)
 if s==n:
  return "yes"
 else :
  return "no"


n=int(input())
print(f1(n))
```

The `f1` function checks if a given number `n` is a perfect number. It calculates the sum of `s` of all its proper positive divisors (numbers that divide `n` evenly, excluding `n` itself). If this sum `s` is equal to `n`, then `n` is a perfect number, and the function returns 'yes'; otherwise, it returns 'no'.

Write a Python function to concatenate any no of dictionaries to create a new one.
Sample Dictionary :
dic1={1:10, 2:20}
dic2={3:30, 4:40}
dic3={5:50,6:60}
Expected Result : {1: 10, 2: 20, 3: 30, 4: 40, 5: 50, 6: 60}

```python
def concat_dicts(*dicts):
    result = {}
    for d in dicts:
        result.update(d)
    return result

dic1 = {1:10, 2:20}
dic2 = {3:30, 4:40}
dic3 = {5:50, 6:60}

print(concat_dicts(dic1, dic2, dic3))
```

The `concat_dicts` function uses `*dicts` to accept an arbitrary number of dictionary arguments. It initializes an empty dictionary `result`. Then, it iterates through each dictionary provided in `dicts` and uses the `update()` method to add all key-value pairs from that dictionary into the `result` dictionary. This effectively merges all input dictionaries into one.

## 1. What are Functions in Python?

Functions are blocks of organized, reusable code that perform a single, related action. They help break our program into smaller, modular chunks. As your programs grow larger, functions make them more organized and manageable. They also promote code reusability, meaning you don't have to write the same code again and again.

### Basic Function Definition

A function is defined using the `def` keyword, followed by the function name, a set of parentheses `()`, and a colon `:`. The code block within the function is indented. The `return` statement is used to send a result back to the caller.
