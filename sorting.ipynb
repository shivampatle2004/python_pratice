```python
l = [5, 2, 8, 1, 3]
for i in range(len(l)):
  for j in range(i+1,len(l)):
    if l[i]>l[j]:
      l[i],l[j]=l[j],l[i]
print(l)
```

```python
l = [5, 2, 8, 1, 3]
for i in range(len(l)):
  m=i
  for j in range(i+1,len(l)):
    if l[j]<l[m]:
      m=j
  l[i],l[m]=l[m],l[i]
print(l)




```
