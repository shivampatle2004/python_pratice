Implement Bubble Sort.

```python
l = [5, 2, 8, 1, 3]

for i in range(len(l)-1):
  f=0
  for j in range(len(l)-i-1):
    if l[j]>l[j+1]:
      f=f+1
      l[j],l[j+1]=l[j+1],l[j]
  if f ==0:
    break
print(l)
```

```python
nums = [64, 25, 12, 22, 11]

n = len(nums)

for i in range(n):
    min_index = i

    for j in range(i + 1, n):
        if nums[j] < nums[min_index]:
            min_index = j

    nums[i], nums[min_index] = nums[min_index], nums[i]

print(nums)
```

```python
nums = [30, 10, 20, 5]

for i in range(1, len(nums)):
    current = nums[i]
    j = i - 1

    while j >= 0 and nums[j] > current:
        nums[j + 1] = nums[j]
        j = j - 1

    nums[j + 1] = current

print(nums)
```
