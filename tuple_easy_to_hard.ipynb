```python
t = (1,2,3,2,4,2,5)
x = 2
c=0
for i in t:
  if i == x:
    c=c+1
print(c)
```

Find Maximum and Minimum Element in a Tuple

```python
t = (7,2,9,1,5)
mx=t[0]
mi=t[0]
for i in t:
  if i > mx:
    mx=i
for i in t:
  if i < mi:
    mi=i
print(mx, mi)


```

Find Duplicate Elements in a Tuple

```python
t = (1,2,3,2,4,5,3,6,2)
t1=()
for i in t :
  if t.count(i)>1:
    t1=t1+(i,)
print(tuple(set(t1)) )
```
