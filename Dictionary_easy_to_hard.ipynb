Q16 — Group Elements by Their First Character

```python
l = ["apple", "ant", "banana", "bat", "cat", "car"]
l=sorted(l)
l2=[]
l3=dict()
for i in range(len(l)) :
  l2.append(l[i][0])
l2=sorted((set(l2)))
for i in range(len(l)):
  for j in l2:
    if l[i][0]==j:
      l3[j]=l[i]
print(l3)





```

```python
l = ["apple", "ant", "banana", "bat", "cat", "car"]

d = {}

for i in l:
    if i[0] not in d:
        d[i[0]] = []

    d[i[0]].append(i)

print(d)
```

```python
Q20 — Create a Nested Dictionary from a List of Tuples

Given:

l = [
    ("A", "Math", 90),
    ("A", "Science", 85),
    ("B", "Math", 78),
    ("B", "Science", 88)
]
```

```python

l = [list(("A", "Math", 90)),
     list(("A", "Science", 85)),
     list(("B", "Math", 78)),
     list(("B", "Science", 88))
]

for i in range(len(l)):
   l[i][1]=l[i][0]+l[i][1]
print(l)
```

```python
l = [
    ("A", "Math", 90),
    ("A", "Science", 85),
    ("B", "Math", 78),
    ("B", "Science", 88)
]
d = {}
for student, subject, score in l:
  if student not in d:
    d[student] = {}
  d[student][subject] = score
print(d)
```
