```python
# 2D dictionary -> JSON
s = {
    'name':'nitish',
     'college':'bit',
     'sem':4,
     'subjects':{
         'dsa':50,
         'maths':67,
         'english':34
     }
}
s
```

```python
# using sequence and dict function
d4 = dict([('name','nitish'),('age',32),(3,3)])
d4
```

```python
# duplicate keys
d5 = {'name':'nitish','name':'rahul'}
d5
```

```python
# mutable items as keys
d6 = {'name':'nitish',(1,2,3):2}
print(d6)
```
