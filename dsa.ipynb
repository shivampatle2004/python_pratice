Find two numbers whose sum equals target

```python
l = [2, 7, 11, 15]
target = 9
for i in range(len(l)):
  for j in range(i+1,len(l)):
    if l[i]+l[j]==target:
      print(l[i],l[j])
      break
```

First Repeating Element

```python
l = [4, 2, 7, 2, 8, 4]
l1=[]
for i in l:
  if i  not in l1:
    l1.append(i)
  else:
    print(i)
    break
```

```python
#non repeteive
l = [4, 5, 1, 2, 1,3, 4, 5]
l1=[]
l2=[]
for i in l:
  if i not in l1:
    l1.append(i)
  else:
    l2.append(i)
print(l1,l2)
for i in l1:
  if i in l2:
    continue
  else:
    print(i)
    break

```
