Remove Duplicates from a List Using a Set While Preserving Order

```python
l = [1,2,2,3,1,4,5,4]
seen=set()

for i in l:
    if i not in seen:
      seen.add(i)

print(seen)
```

Find All Unique Pairs Whose Sum Equals Target

```python
from itertools import combinations

l = [1,2,3,4,5,6]
target = 7
for i in range(len(l)+1):
  for j in combinations(l,i):
    if sum(j)==target:
      print(j)
```
