Find the Union of Two Sets

```python
s1 = {1,2,3,4}
s2 = {3,4,5,6}
print(s1 | s2)
#
print(s1.union(s2))
#
print(s1)
(s1.update(s2))
print(s1)
```

Find the Intersection of Two Sets

```python
s1 = {1,2,3,4}
s2 = {3,4,5,6}
print(s1 & s2)
print(s1.intersection(s2))
s1.intersection_update(s2)
print(s1)
```

Find the Difference Between Two Sets

```python
s1 = {1,2,3,4}
s2 = {3,4,5,6}
print(s1-s2)
print(s2-s1)
print(s1.difference(s2))
s1.difference_update(s2)
print(s1)
```

Find the Symmetric Difference Between Two Sets

```python
s1 = {1,2,3,4}
s2 = {3,4,5,6}
print(s1^s2)
print(s1.symmetric_difference(s2))
s1.symmetric_difference_update(s2)
print(s1)

```

Remove Duplicates from a List Using a Set While Preserving Order

```python
l = [1,2,2,3,1,4,5,4]
seen=set()

for i in l:
    if i not in seen:
      seen.add(i)

print(seen)
```

Find All Unique Pairs Whose Sum Equals Target

```python
from itertools import combinations

l = [1,2,3,4,5,6]
target = 7
for i in range(len(l)+1):
  for j in combinations(l,i):
    if sum(j)==target:
      print(j)
```
