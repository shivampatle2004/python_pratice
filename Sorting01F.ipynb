```python
def bubble_sort(arr):
  n = len(arr)
  for i in range(n):
    swapped = False
    for j in range(0, n - i - 1):
      if arr[j] > arr[j + 1]:
        arr[j], arr[j + 1] = arr[j + 1], arr[j]
        swapped = True
    if not swapped:
      break
  return arr


if __name__ == "__main__":
  print(bubble_sort([64, 34, 25, 12, 22, 11, 90]))
```

```python
def selection_sort(arr):
  n = len(arr)
  for i in range(n):
    min_idx = i
    for j in range(i + 1, n):
      if arr[j] < arr[min_idx]:
        min_idx = j
    arr[i], arr[min_idx] = arr[min_idx], arr[i]
  return arr


if __name__ == "__main__":
  print(selection_sort([29, 10, 14, 37, 13]))
```

```python
def insertion_sort(arr):
  for i in range(1, len(arr)):
    key = arr[i]
    j = i - 1
    while j >= 0 and key < arr[j]:
      arr[j + 1] = arr[j]
      j -= 1
    arr[j + 1] = key
  return arr


if __name__ == "__main__":
  print(insertion_sort([12, 11, 13, 5, 6]))
```

```python
def merge_sort(arr):
  if len(arr) <= 1:
    return arr

  mid = len(arr) // 2
  left = merge_sort(arr[:mid])
  right = merge_sort(arr[mid:])

  return merge(left, right)


def merge(left, right):
  result = []
  i = j = 0
  while i < len(left) and j < len(right):
    if left[i] < right[j]:
      result.append(left[i])
      i += 1
    else:
      result.append(right[j])
      j += 1
  result.extend(left[i:])
  result.extend(right[j:])
  return result


if __name__ == "__main__":
  print(merge_sort([38, 27, 43, 3, 9, 82, 10]))
```

```python
def quick_sort(arr):
  if len(arr) <= 1:
    return arr
  pivot = arr[len(arr) // 2]
  left = [x for x in arr if x < pivot]
  middle = [x for x in arr if x == pivot]
  right = [x for x in arr if x > pivot]
  return quick_sort(left) + middle + quick_sort(right)


if __name__ == "__main__":
  print(quick_sort([10, 7, 8, 9, 1, 5]))
```

```python
def sort_colors(nums):
  low, mid, high = 0, 0, len(nums) - 1
  while mid <= high:
    if nums[mid] == 0:
      nums[low], nums[mid] = nums[mid], nums[low]
      low += 1
      mid += 1
    elif nums[mid] == 1:
      mid += 1
    else:
      nums[mid], nums[high] = nums[high], nums[mid]
      high -= 1
  return nums


if __name__ == "__main__":
  print(sort_colors([2, 0, 2, 1, 1, 0]))
```

```python
def merge_intervals(intervals):
  if not intervals:
    return []

  intervals.sort(key=lambda x: x[0])
  merged = [intervals[0]]

  for current in intervals[1:]:
    prev = merged[-1]
    if current[0] <= prev[1]:
      prev[1] = max(prev[1], current[1])
    else:
      merged.append(current)

  return merged


if __name__ == "__main__":
  print(merge_intervals([[1, 3], [2, 6], [8, 10], [15, 18]]))
```

```python
def find_kth_largest(nums, k):
  nums.sort(reverse=True)
  return nums[k - 1]


if __name__ == "__main__":
  print(find_kth_largest([3, 2, 1, 5, 6, 4], 2))
```

```python
def is_anagram(s, t):
  if len(s) != len(t):
    return False
  return sorted(s) == sorted(t)


if __name__ == "__main__":
  print(is_anagram("anagram", "nagaram"))
```

```python
from functools import cmp_to_key


def largest_number(nums):
  str_nums = list(map(str, nums))

  def compare(x, y):
    if x + y > y + x:
      return -1
    elif x + y < y + x:
      return 1
    return 0

  str_nums.sort(key=cmp_to_key(compare))
  result = "".join(str_nums)
  return "0" if result[0] == "0" else result


if __name__ == "__main__":
  print(largest_number([3, 30, 34, 5, 9]))
```


Problem: The Gravity-Flip Relay Sort

You are given an array of N distinct integers representing magnetic blocks sitting on a vertical track. You do not have standard swap operations. Instead, you have a Magnetic Flipper that performs a Gravity-Flip of length K (where 1 ≤ K ≤ N):
1. Flip: Pick any contiguous subarray of length K and reverse its order.
2. Gravity Settle: Immediately after the flip, gravity pulls down inside that same subarray—the smallest element in that subarray sinks straight to the bottom (the left-most index of the subarray), shifting all other elements in that subarray one spot to the right.

The Twist (Cooldown Constraint)

• Every time you use a specific length K, that exact length goes on cooldown and cannot be used in the immediately following move (though you can reuse it after making at least one move of a different length).

Objective

Find the minimum number of Gravity-Flip moves required to sort the entire array in strictly ascending order, or return -1 if it is impossible.

Example Walkthrough

• Input: arr = [3, 1, 4, 2]
• Move 1 (Choose K = 3 on indices 0..2 [3, 1, 4]):
	• Flip: [4, 1, 3]
	• Gravity Settle (smallest is 1, sinks to index 0): [1, 4, 3]
	• Full Array: [1, 4, 3, 2] (Note: K = 3 is now on cooldown for Move 2)
• Move 2 (Choose K = 2 on indices 2..3 [3, 2]):
	• Flip: [2, 3]
	• Gravity Settle (smallest is 2, sinks to index 2): [2, 3]
	• Full Array: [1, 4, 2, 3] (Note: K = 2 is now on cooldown; K = 3 is available again)
• Move 3 (Choose K = 3 on indices 1..3 [4, 2, 3]):
	• Flip: [3, 2, 4]
	• Gravity Settle (smallest is 2, sinks to index 1): [2, 3, 4]
	• Full Array: [1, 2, 3, 4] — Sorted!
• Output: 3

```python
from collections import deque

def min_gravity_flips(arr: list[int]) -> int:
    n = len(arr)
    target = tuple(sorted(arr))
    start = tuple(arr)

    if start == target:
        return 0

    queue = deque([(start, -1, 0)])
    visited = {(start, -1)}

    while queue:
        curr, last_k, moves = queue.popleft()

        for k in range(1, n + 1):
            if k == last_k:
                continue

            for i in range(n - k + 1):
                sub = list(curr[i : i + k])
                sub.reverse()

                min_val = min(sub)
                sub.remove(min_val)
                settled_sub = [min_val] + sub

                next_state = curr[:i] + tuple(settled_sub) + curr[i + k:]

                if next_state == target:
                    return moves + 1

                if (next_state, k) not in visited:
                    visited.add((next_state, k))
                    queue.append((next_state, k, moves + 1))

    return -1

arr = [3, 4, 1, 2]
print("Minimum moves:", min_gravity_flips(arr))
```
