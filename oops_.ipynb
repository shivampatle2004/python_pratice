````python
class LessError(Exception):
  """Raised when the input is less than the number"""
  pass

class GreaterError(Exception):
  """Raised when the input is greater than the number"""
  pass

number = 10
while True:
  try:
    i = int(input("Guess the number: "))
    if i < number:
      raise LessError
    elif i > number:
      raise GreaterError

    break # Exit the loop if the guess is correct
  except LessError:
    print("Your guess is too low!")
  except GreaterError:
    print("Your guess is too high!")
  except ValueError:
    print("Invalid input. Please enter an integer.")

print("Congratulations! You guessed the number.")
````

### Explanation of the Guessing Game Logic

*   **Custom Exceptions**: `class LessError(Exception): pass` allows us to define our own specific error types. This is useful for signaling specific conditions (like a low guess) that aren't built-in Python errors.
*   **The `try` block**: This contains the code that might cause an issue or where we want to check conditions.
*   **The `raise` keyword**: When `i < number` or `i > number` is true, we manually trigger an exception to jump straight to the error handling section.
*   **The `except` blocks**: These 'catch' the errors. Instead of the program crashing, it prints a message and continues to the next iteration of the loop.
*   **Flow Control**: The `break` statement only runs if no exceptions are raised, meaning the user successfully guessed the number `10`.

### Understanding the Exception Syntax

1.  **Triple Quotes (`""" ... """`)**: These are called **Docstrings**. In Python, they are used to document what a class or function does. They don't affect how the code runs, but they help other programmers (or you in the future) understand the purpose of the code.

2.  **The `pass` keyword**: In Python, you cannot have an empty class or function. If you want to define a class but don't need to add any new methods or logic to it, you must use `pass`. It literally means "do nothing."

3.  **Why use them here?**: We are creating `LessError` just so we have a unique 'name' to catch in our `try/except` block. Since the standard `Exception` class already has all the functionality we need, we just name our new class and use `pass` to keep the syntax valid.

### 📚 OOP Theory Revision: The 4 Pillars

Object-Oriented Programming (OOP) is based on four main concepts. Here is how they apply to our code:

#### 1. Encapsulation (Grouping Data & Logic)
*   **Definition**: Bundling data (variables) and the methods that work on that data into a single unit (a class).
*   **In our code**: The `GuessingGame` class encapsulates the `secret_number` and the `play()` logic. You don't see the internal variable `self.number` from outside; you just call `.play()`.

#### 2. Abstraction (Hiding Complexity)
*   **Definition**: Hiding the complex background details and only showing the essential features to the user.
*   **In our code**: When you run `my_game.play()`, you don't need to know how the `while` loop or the `try-except` blocks work. You just know that calling `play()` starts the game.

#### 3. Inheritance (Reusing Code)
*   **Definition**: Allowing a class (child) to derive properties and methods from another class (parent).
*   **In our code**: `class LessError(Exception): pass`. Here, `LessError` **inherits** from Python's built-in `Exception` class. It gets all the functionality of an error without us writing any extra code.

#### 4. Polymorphism (Multiple Forms)
*   **Definition**: The ability of different classes to be treated as instances of the same general class through the same interface.
*   **In our code**: Both `LessError` and `GreaterError` are treated as `Exceptions`. The `try` block doesn't care which specific error it is; it just looks for a match in the `except` blocks. In a larger system, you could have different types of games (Number guess, Word guess) that both have a `.play()` method.

### 🏗️ Fundamental Building Blocks of OOP

To master OOP, you need to understand these basic terms and how they work together:

#### 1. Class
A **Class** is a blueprint or a template. It defines what the objects will look like and what they can do.
*   *Example*: A `Car` blueprint defines that every car has a color and can drive.

#### 2. Object (Instance)
An **Object** is the actual thing created from the blueprint.
*   *Example*: If `Car` is the class, then your specific red Tesla is an **Instance** of that class.

#### 3. Attributes (Data/Variables)
These are the variables stored inside a class. They represent the **state** or characteristics of an object.
*   *Example*: `self.color = "Red"` or `self.speed = 0`.

#### 4. Methods (Behavior/Functions)
Methods are functions defined inside a class. They represent the **actions** an object can take.
*   *Example*: `def drive(self):` or `def stop(self):`.

#### 5. The Constructor (`__init__`)
This is a special method that runs automatically when you create a new object. Its main job is to initialize the attributes.
*   *Example*:
```python
class Dog:
    def __init__(self, name):
        self.name = name # Assigning the name immediately
```

#### 6. `self` Keyword
`self` represents the specific instance of the object. It allows the code to distinguish between "my name" and "your name" when multiple objects are created from the same class.

````python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def move(self):
        pass

class Human(Animal):
    def move(self):
        print("I can walk and run")

class Lion(Animal):
    def move(self):
        print("I can roar and pounce")

h = Human()
h.move()

l = Lion()
l.move()
````

````python
class Rect:
    def __init__(self, length, breadth):
        self.__length = length   # Private attribute
        self.__breadth = breadth # Private attribute

    def display(self):
        print(f"Length: {self.__length}")
        print(f"Breadth: {self.__breadth}")

# Create object
rect = Rect(10, 5)

# This would cause an error: print(rect.__length)
# Instead, we use a public method to see the private data:
rect.display()
````

````python
class Parent:
    def __init__(self):
        # Protected member
        self._variable = "I am protected"

class Child(Parent):
    def __init__(self):
        Parent.__init__(self)
        print("Accessing protected member in child class:", self._variable)

# Create object of Child class
obj = Child()

# Accessing protected member directly (possible, but discouraged by convention)
print("Accessing protected member outside:", obj._variable)
````

### 🛡️ Protected Access Modifier Example

In this example:
1. `detail` is the parent class with protected attributes (prefixed with `_`).
2. `pro_mod` is the child class that accesses those protected attributes.

````python
class detail:
    def __init__(self, name, age, job):
        # Protected members
        self._name = name
        self._age = age
        self._job = job

class pro_mod(detail):
    def __init__(self, name, age, job):
        # Initialize parent class
        super().__init__(name, age, job)

    def display_info(self):
        # Accessing protected members from the parent class
        print(f"Name: {self._name}")
        print(f"Age: {self._age}")
        print(f"Job: {self._job}")

# Instantiate the child class
person = pro_mod("Rahul", 28, "Developer")

# Call method that uses protected members
person.display_info()
````

````python
class BankAcc:
    def __init__(self, balance):
        self.balance = balance
        print(f"Account created with balance: {self.balance}")

    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited: {amount}. New Balance: {self.balance}")

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print(f"Withdrew: {amount}. New Balance: {self.balance}")
        else:
            print("Insufficient funds!")

# Creating an account with 1000
my_acc = BankAcc(1000)
my_acc.deposit(500)
my_acc.withdraw(200)
````

### 💡 The "Self" Analogy

Imagine a **House Blueprint** (The Class). It says every house has a `front_door_color`.

When you build two real houses (Objects):
*   **House A** (self) sets its `color` to Red.
*   **House B** (self) sets its `color` to Blue.

In Python, we write `self.color = color` because:
1.  **`self.color`**: This is the actual wall of the house you are standing in.
2.  **`color`**: This is just the bucket of paint you brought with you to the construction site.

````python
class Robot:
    def __init__(self, name):
        # 'self' ensures the name sticks to THIS specific robot
        self.name = name

    def say_hi(self):
        # Without 'self', the robot wouldn't remember its own name!
        print(f"Hello, my name is {self.name}")

# Creating two different robots
bob = Robot("Bob-Bot")
alice = Robot("Alice-Bot")

bob.say_hi()   # 'self' here is bob
alice.say_hi() # 'self' here is alice
````

### 🚀 Step-by-Step: Class, Object, and Operations

Think of a **Calculator**:
1.  **The Class**: Defines that every calculator can add numbers.
2.  **The Object**: Your physical calculator on your desk.
3.  **The Operation**: Pressing the '+' button.

````python
# STEP 1: Create the Blueprint (Class)
class SimpleCalc:
    def __init__(self, brand):
        self.brand = brand  # This saves the brand name to the object

    # STEP 3: Define Operations (Methods)
    def add(self, num1, num2):
        result = num1 + num2
        print(f"[{self.brand}] {num1} + {num2} = {result}")

# STEP 2: Create the Object (Instance)
my_calculator = SimpleCalc("Casio")

# Use the Object to perform operations
my_calculator.add(10, 5)
my_calculator.add(100, 200)
````

````python
class Student:
    def __init__(self, name, mark):
        self.name = name
        # Logic: If mark is negative, set it to 0
        if mark < 0:
            self.mark = 0
        else:
            self.mark = mark

    def display_info(self):
        print(f"Student: {self.name} | Mark: {self.mark}")

# Creating objects with names and marks
s1 = Student("Arjun", 85)
s1.display_info()

s2 = Student("Sara", -10)
s2.display_info()
````

### ✨ Magic Methods (`__`)

Magic methods allow your objects to behave like built-in Python types. For example, we can tell Python exactly how to describe a `Student` when we use `print()`.

````python
class Student:
    def __init__(self, name, mark):
        self.name = name
        self.mark = mark if mark >= 0 else 0

    # This 'Magic Method' defines what happens when you print the object
    def __str__(self):
        return f"[Student Object] Name: {self.name}, Score: {self.mark}"

s3 = Student("Arjun", 95)

# Python automatically calls s3.__str__() here!
print(s3)
````

### 🔒 Private Variables (`__variable`)

In Python, `__` (double underscore) is used to make a variable **Private**. This prevents accidental changes to sensitive data from outside the class.

````python
class User:
    def __init__(self, username, pin):
        self.username = username  # Public: anyone can see this
        self.__pin = pin          # Private: hidden from the outside

    def check_pin(self, entered_pin):
        if self.__pin == entered_pin:
            print("Access Granted!")
        else:
            print("Access Denied!")

# Create a user
u = User("Arjun", 1234)

print(f"Username: {u.username}")

# This line would cause an ERROR because __pin is hidden:
# print(u.__pin)

# We must use the class method to interact with the private data
u.check_pin(1234)
````
