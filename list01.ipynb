Write a program to add two lists index-wise. Create a new list that contains the 0th index item from both the list, then the 1st index item, and so on till the last element. any leftover items will get added at the end of the new list.

````python
list1 = ["M", "na", "i", "Kh"]
list2 = ["y", "me", "s", "an"]
list4=[]
for i in range(0,len(list1)):
   list3=[]

   list3.append(list1[i])
   list3.append(list2[i])
   list4.append(list3)
   del list3
print(list4)
````

Running Sum on list
Write a program to print a list after performing running sum on it.

i.e:

Input:

list1 = [1,2,3,4,5,6]
Output:

[1,3,6,10,15,21]

````python
list1 = [1,2,3,4,5,6]
list2=[]
sum=0
for i in list1:
  sum=sum+i
  list2.append(sum)
print(list2)


````

You are given a list of integers. You are asked to make a list by running through elements of the list by adding all elements greater and itself.
i.e. Say given list is [2,4,6,10,1] resultant list will be [22,20,10,23].

For 1st element 2 ->> these are greater (4+6+10) values and 2 itself so on adding becomes 22.

For 2nd element 4 ->> greater elements are (6, 10) and 4 itself, so on adding 20

like wise for all other elememts.

[2,4,6,10,1]-->[22,20,16,10,23]

````python
l= [2,4,6,10,1]
l1=[]
for i in range(0,len(l)):
   s=0
   for j in range(0,len(l)):
    if(l[j]>=l[i]):

      s=s+l[j]

   l1.append(s)
print(l1)



````

Sort a list of alphanumeric strings based on product value of numeric character in it. If in any string there is no numeric character take it's product value as 1.
Input:

['1ac21', '23fg', '456', '098d','1','kls']
Output:

['456', '23fg', '1ac21', '1', 'kls', '098d']

````python
l1=['1ac21', '23fg', '456', '098d','1','kls']
l2=[]
r=''
r1=''

# Step 1: Store product values in l2
for i in l1:
    s1=1
    s2=0
    r=i

    for j in r:
        r1=j
        if r1.isdigit():
            s1 = s1 * int(r1)
            s2 = 1

    if s2==0:
        s1=1

    l2.append(s1)

# Step 2: Manual sorting (Descending order)
n = len(l2)

for i in range(n):
    for j in range(i+1, n):
        if l2[i] < l2[j]:   # descending
            # swap product values
            l2[i], l2[j] = l2[j], l2[i]

            # swap corresponding strings
            l1[i], l1[j] = l1[j], l1[i]

print(l1)
````
