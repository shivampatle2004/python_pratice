```python
l = [3, 5, 2, 6, 4]
l1=[]
for i  in range(len(l)):
  for j in range(i+1,len(l)) :
    if l[i] == l[j]:
      continue
    else:
      l1.append(l[i]*l[j])
print(max(l1))

```
