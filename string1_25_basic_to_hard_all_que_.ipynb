Q1. Reverse a String

You are given a string S. Return the reversed string.

Input:
A single string S (may contain letters, digits, spaces)

Output:
Print the reversed string.

Example:
Input: hello
Output: olleh


```python
s = input()
rev = ""
for i in range(len(s)-1, -1, -1):
    rev = rev + s[i]
print(rev)
```

```python
s = input()
rev = []

for i in range(len(s)-1, -1, -1):
    rev.append(s[i])

print("".join(rev))
```

Q13: Remove Duplicate Characters from String


```python
s=input()
r=""
s1=list(s)
i = 0
while i < len(s1):
  j = i + 1
  while j < len(s1):
    if s1[i]==s1[j]:
      del s1[j]
    else:
      j += 1
  r=r+s1[i]
  i += 1
print(r)
```

The core of your question lies in how `j` is handled within the inner `while` loop, especially when `del s1[j]` occurs. Let's break it down:

**Outer Loop (`while i < len(s1)`):**
This loop iterates through the list `s1` from the first element to the last. `i` represents the index of the character we are currently checking against all subsequent characters to remove duplicates.

**Inner Loop (`while j < len(s1)`):**
For each character `s1[i]` from the outer loop, this inner loop starts checking from the next character (`j = i + 1`) to the end of the list.

**Conditional Increment of `j`:**

*   **`if s1[i] == s1[j]` (Duplicate found):**
    If a duplicate is found (`s1[i]` is the same as `s1[j]`), `del s1[j]` is executed. This removes the element at index `j` from the list `s1`. **Crucially, when an element is deleted, all subsequent elements in the list shift one position to the left.** For example, if `s1` was `[a, b, c, d]` and you `del s1[1]` (which is 'b'), `s1` becomes `[a, c, d]`. The character 'c' which was at index `2` is now at index `1`. In this case, `j` is *not* incremented.

    By *not* incrementing `j` after a deletion, the next iteration of the inner `while` loop re-evaluates `s1[j]`. This is correct because the element that was *previously* at `j+1` has now moved to index `j`, and we still need to compare it with `s1[i]`. If `j` were incremented, you would skip checking the new element that shifted into position `j`.

*   **`else: j += 1` (No duplicate):**
    If `s1[i]` is not equal to `s1[j]`, it means `s1[j]` is a unique character (relative to `s1[i]`). In this case, `j` is incremented to move to the next character in the list to continue the comparison.

**Why it doesn't lead to an infinite loop:**

The inner `while` loop condition is `j < len(s1)`. This condition will eventually become false for two reasons:

1.  **`j` increments:** In the `else` block (most common case), `j` directly increases, moving towards `len(s1)`.
2.  **`len(s1)` decreases:** In the `if` block, `del s1[j]` reduces the length of `s1`. Even if `j` doesn't increment, `len(s1)` gets smaller, effectively

Here's a trace with the input `programming` to clarify:

`s = "programming"`
`s1 = ['p', 'r', 'o', 'g', 'r', 'a', 'm', 'm', 'i', 'n', 'g']`
`r = ""`

**1. `i = 0`, `s1[i] = 'p'`**
   - `j` iterates from 1 to 10. No 'p' duplicates are found. `s1` remains unchanged.
   - `r = "p"`

**2. `i = 1`, `s1[i] = 'r'`**
   - `j = 2`, `s1[2] = 'o'` (not 'r'), `j` becomes 3.
   - `j = 3`, `s1[3] = 'g'` (not 'r'), `j` becomes 4.
   - `j = 4`, `s1[4] = 'r'` (is 'r')
     - `del s1[4]` (the second 'r' is removed). `s1` becomes `['p', 'r', 'o', 'g', 'a', 'm', 'm', 'i', 'n', 'g']`.
     - `j` remains 4. (`len(s1)` is now 10)
   - `j = 4`, `s1[4]` is now `'a'` (not 'r'), `j` becomes 5.
   - `j` continues to 9. No more 'r' duplicates. The inner loop ends when `j` is 10 (as `j < len(s1)` (10) is false).
   - `r = "pr"`

**3. `i = 2`, `s1[i] = 'o'`**
   - `j` iterates from 3 to 9. No 'o' duplicates found. `s1` remains unchanged.
   - `r = "pro"`

**4. `i = 3`, `s1[i] = 'g'`**
   - `j` iterates. When `j = 9`, `s1[9]` is `'g'` (is 'g')
     - `del s1[9]` (the second 'g' is removed). `s1` becomes `['p', 'r', 'o', 'g', 'a', 'm', 'm', 'i', 'n']`.
     - `j` remains 9. (`len(s1)` is now 9)
   - Now, `j` (9) is no longer less than `len(s1)` (9), so the inner loop ends.
   - `r = "prog"`

**5. `i = 4`, `s1[i] = 'a'`**
   - `j` iterates from 5 to 8. No 'a' duplicates found. `s1` remains unchanged.
   - `r = "proga"`

**6. `i = 5`, `s1[i] = 'm'`**
   - `j` iterates. When `j = 6`, `s1[6]` is `'m'` (is 'm')
     - `del s1[6]` (the second 'm' is removed). `s1` becomes `['p', 'r', 'o', 'g', 'a', 'm', 'i', 'n']`.
     - `j` remains 6. (`len(s1)` is now 8)
   - `j = 6`, `s1[6]` is now `'i'` (not 'm'), `j` becomes 7.
   - `j = 7`, `s1[7]` is `'n'` (not 'm'), `j` becomes 8.
   - Now, `j` (8) is no longer less than `len(s1)` (8), so the inner loop ends.
   - `r = "progam"`

**7. `i = 6`, `s1[i] = 'i'`**
   - `j` iterates. No 'i' duplicates found.
   - `r = "progami"`

**8. `i = 7`, `s1[i] = 'n'`**
   - `j` iterates. No 'n' duplicates found.
   - `r = "progamin"`

**9. `i = 8`**
   - `i` (8) is no longer less than `len(s1)` (8). The outer loop ends.

**Final output:** `progamin`

This demonstrates that the conditional increment of `j` combined with the dynamic change in `len(s1)` due to deletions correctly handles the removal of duplicates without an infinite loop.
