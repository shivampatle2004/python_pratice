Q1. Reverse a String

You are given a string S. Return the reversed string.

Input:
A single string S (may contain letters, digits, spaces)

Output:
Print the reversed string.

Example:
Input: hello
Output: olleh


```python
s = input()
rev = ""
for i in range(len(s)-1, -1, -1):
    rev = rev + s[i]
print(rev)
```

```python
s = input()
rev = []

for i in range(len(s)-1, -1, -1):
    rev.append(s[i])

print("".join(rev))
```

Q2. Palindrome Check

Given a string S, determine whether it is a palindrome.

Condition: Ignore case and spaces.

Output:
Print "YES" if palindrome, otherwise "NO"

Example:
Input: Madam
Output: YES

```python
s=input()
s=s.lower()
s=s.split()
print(s)
s="".join(s)
print(s)
s1=s[::-1]
if s==s1:
  print("yes")
else :
  print("no")
```

```python
s = input().lower().replace(" ", "")

if s == s[::-1]:
    print("YES")
else:
    print("NO")
```

Manual Case Conversion
Example 1
Input: AbC
Output: aBc

```python
s=input()
print(s.swapcase())
```

```python

s = input()
result = ""

for ch in s:
    if 'a' <= ch <= 'z':
        result = result + chr(ord(ch) - 32)
    elif 'A' <= ch <= 'Z':
        result = result + chr(ord(ch) + 32)
    else:
       result = result + ch

print(result)
#chr() number ascii value to char
#ord() alfa to asci digit
```

Q9: Anagram Check (Interview Style)
📌 Problem

You are given two strings S1 and S2.
Check whether they are anagrams of each other.

🧠 What is an anagram?

Two strings are anagrams if:
👉 They contain the same characters with the same frequency
👉 Order does NOT matter

📥 Input
First line: string S1
Second line: string S2
📤 Output

Print:

"YES" → if anagram
"NO" → otherwise
🧪 Examples
Example 1

Input:

listen
silent

Output:

YES


```python
#first approch
s=input()
s1=input()
s=sorted(s)
s1=sorted(s1)
if s==s1:
  print("yes")
else:
  print("no")
```

```python
s1 = input().lower().replace(" ", "")
s2 = input().lower().replace(" ", "")

if len(s1) != len(s2):
    print("NO")
else:
    for ch in s1:
        if s1.count(ch) != s2.count(ch):
            print("NO")
            break
    else:
        print("YES")
```
